```python
#import all required Modules
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# install The Titanic data from seaborn
print("=" * 60)
print("1. LOADING DATASET")
print("=" * 60)
data = sns.load_dataset("titanic")
```

```python
#converting data into the dataframe
df = pd.DataFrame(data)
```

```python
print("\nDataset Information:")
df.info()
```

```python
# Shape
df.shape
```

```python
# Previewing the data fromm top to bottom
df.head()
```

```python
# Previewing the data from bottom to top
df.tail()
```

```python
# Listing down the columns
df.columns.values
```

```python
#using info to see null as well column data type
df.info()
```

```python
#checking total null value in the data
df.isnull().sum()
```

```python
# - 'embarked': Fill missing values with the most frequent port (mode)
embarked_mode = df["embarked"].mode()[0]
df["embarked"] = df["embarked"].fillna(embarked_mode)
```

```python
#droping the data isna value because it is very low
df["embarked"].isna().dropna()
df["embarked"]
```

```python
# Imputing missing values for embarked

# finding the most appeared value in embarked column

df['deck'].value_counts()

#C deck is more so we will fill C in Isna value in deck
df["deck"].fillna("C",inplace=True)


#We fill Age Column with the mean age
df["age"].fillna(df["age"].mean(),inplace=True)
```

```python
#checking total null value in the data
#Now we Solve the Problem of the Null Value
df.isnull().sum()
```

```python
#Changing all the Other type data into required Categoryical Type data
print("\n" + "=" * 60)
print("3. DATA TYPE CONVERSION")
print("=" * 60)
df['survived']=df['survived'].astype('category')
df['pclass']=df['pclass'].astype('category')
df['sex']=df['sex'].astype('category')
df['age']=df['age'].astype('int')
df['embarked']=df['embarked'].astype('category')

#checking type agaain
df.info()
```

```python
#checking Statistical Value 
print("\n" + "=" * 60)
print("4. DESCRIPTIVE STATISTICS")
print("=" * 60)
df.describe().T
```

```python
## changing the theme of the chart
print("""
# ==========================================
# MODERN & ATTRACTIVE STYLING CONFIGURATION
# ==========================================
""")
sns.set_theme(style="ticks", palette="viridis")
```

## Bivariate Analysis

```python
print("\n" + "=" * 60)
print("5. UNIVARIATE & BIVARIATE VISUALIZATIONS")
print("=" * 60)
```

```python
# Histogram of passenger ages Histograms are perfect for continuous variables. This shows how passenger ages are distributed.
#  Most passengers were between 20–40 years old.
df['age'].plot(kind='hist', bins=30, edgecolor='black', color='green')

plt.title("Age Distribution (Histogram)")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# Vertical countplot
sns.countplot(x='survived', data=df)
plt.title("Survival Count")
plt.xlabel("Survived (0 = Died, 1 = Survived)")
plt.ylabel("Count")
plt.show()

# Death percentage calculation
death_percent = round((df['survived'].value_counts().values[0] / 891) * 100)
print("Out of 891, {}% people died in the accident".format(death_percent))
```

```python
# Percentage distribution of Pclass
print((df['pclass'].value_counts() / 891) * 100)

# Vertical countplot
sns.countplot(x='pclass', data=df)
plt.title("Passenger Class Distribution")
plt.xlabel("Pclass (1 = Upper, 2 = Middle, 3 = Lower)")
plt.ylabel("Count")
plt.show()
```

```python
# Percentage distribution of Sex
print((df['sex'].value_counts() / 891) * 100)

# Vertical countplot
sns.countplot(x='sex', data=df)
plt.title("Passenger Gender Distribution")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.show()
```

```python
# Percentage distribution
print((df['sibsp'].value_counts() / 891) * 100)

# Vertical countplot
sns.countplot(x='sibsp', data=df)
plt.title("Number of Siblings/Spouses Aboard")
plt.xlabel("SibSp (siblings/spouses)")
plt.ylabel("Count")
plt.show()
```

```python
# Percentage distribution
print((df['parch'].value_counts() / 891) * 100)

# Vertical countplot
sns.countplot(x='parch', data=df)
plt.title("Number of Parents/Children Aboard")
plt.xlabel("Parch (parents/children)")
plt.ylabel("Count")
plt.show()
```

```python

# Correct countplot
sns.countplot(x='embarked', data=df)
plt.title("Port of Embarkation Distribution")
plt.xlabel("Embarked (C = Cherbourg, Q = Queenstown, S = Southampton)")
plt.ylabel("Count")
plt.show()
```

```python
#Making chart of box plot for Check the median age
# Boxplot of Age
sns.boxplot(x='age', data=df)   # vertical
plt.title("Age Distribution with Median")
plt.xlabel("Age")
plt.show()
```

```python
sns.distplot(df['age'], bins=30, kde=True)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Density")
plt.show()
```

# Multivariate Analysis

```python
# Survival vs Gender
sns.countplot(x='sex', hue='survived', data=df)
plt.title("Survival by Gender")
plt.xlabel("Sex")
plt.ylabel("Count")
plt.show()
```

```python
# Age Distribution by Survival
sns.boxplot(x='survived', y='age', data=df)
plt.title("Age Distribution by Survival")
plt.xlabel("Survived (0 = Died, 1 = Survived)")
plt.ylabel("Age")
plt.show()
```

```python
# Higher fares (wealthier passengers) often correlate with survival.
sns.countplot(x='embarked', hue='survived', data=df)
plt.title("Survival by Embarkation Port")
plt.xlabel("Embarked (C = Cherbourg, Q = Queenstown, S = Southampton)")
plt.ylabel("Count")
plt.show()
```

```python
# Scatter plots show relationships between two continuous variables. Here, it compares age and fare, with survival indicated by color.
# You can see wealthier passengers (higher fares) had better survival chances.
# Scatter plot of Age vs Fare, colored by survival
plt.scatter(df['age'], df['fare'], c=df['survived'], cmap='coolwarm', alpha=0.6)

plt.title("Age vs Fare (Colored by Survival)")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.colorbar(label="Survived (0 = No, 1 = Yes)")
plt.show()
```

### ✨ Conclusion:  
Multivariate analysis reveals patterns of survival — women and children, higher-class passengers, and those who paid higher fares had better chances.

```python
# Survival by Passenger Class (Grouped Bar Chart)  using matplotlib

survival_by_class = df.groupby(['pclass', 'survived']).size().unstack()

survival_by_class.plot(kind='bar', stacked=False)
plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")
plt.legend(["Died", "Survived"])
plt.show()
```

```python
# Age vs Fare (Scatter Plot)
plt.figure(figsize=(8,6))
plt.scatter(df['age'], df['fare'], c=df['survived'], cmap='coolwarm', alpha=0.6)
plt.title("Age vs Fare (Colored by Survival)")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.colorbar(label="Survived (0 = No, 1 = Yes)")
plt.show()
```

```python
# Correlation matrix for numeric features
corr = df[['age','fare','sibsp','parch','survived']].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Heatmap")
```

```python
# Figure 5: Passenger Class Distribution (Pie Chart)
plt.figure(figsize=(7, 7))
class_counts = df["class"].value_counts()
explode_values = (0.1, 0, 0)
plt.pie(
    class_counts,
    labels=class_counts.index,
    autopct="%1.1f%%",
    startangle=140,
    explode = explode_values,
    colors=sns.color_palette("pastel"),
    shadow=True,
    
    wedgeprops={"edgecolor": "black", "linewidth": 1},
)
plt.title("Passenger Proportion by Class", fontsize=14, weight="bold")
plt.show()
```

 #### The heatmap shows the correlation between numerical features in the Titanic dataset.

Correlation values range from -1 (strong negative) to +1 (strong positive).

Color meaning:

Red → positive correlation

Blue → negative correlation

White/neutral → near zero correlation
